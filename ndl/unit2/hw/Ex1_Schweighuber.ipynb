```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from pyMLaux import plot_2d_prediction
```

```python
# Initial data from ex

X = np.array([[0.1, 0.5], [0.2, 0.9], [0.4, 0.7], [0.5, 0.8],
              [0.4, 0.2], [0.6, 0.3], [0.7, 0.05], [0.8, 0.5]])
y = np.array([1, 1, 1, 1, -1, -1, -1, -1])  # ADALINE targets: {-1, +1}


w = np.array([1.0, 1.0])
theta_init = -1.2
eta = 0.3
maxiter = 100
```

## Exercise implementation without theta as weight

```python
def adaline_activation_ref(x, w, theta):
    return x.dot(w) - theta

def adaline_learning_ref(X, y, w, theta, eta, maxiter=100):
    w = w.copy().astype(float)
    theta = float(theta)
    
    for epoch in range(maxiter):
        for i in range(len(X)):
            output = adaline_activation_ref(X[i], w, theta)
            error = y[i] - output
            w += eta * error * X[i]
            theta -= eta * error
            
    return w, theta

w_ref, theta_ref = adaline_learning_ref(X, y, w, theta_init, eta, maxiter)

print(f"Final w:     [{w_ref[0]:.8f}, {w_ref[1]:.8f}]")
print(f"Final theta: {theta_ref:.8f}")
```

## Extension with theta as weight

```python
def adaline_activation_ext(x_ext, w_ext):
    # theta is now part of w_ext
    return np.dot(x_ext, w_ext)

def adaline_learning_weights(X, y, w_ext, eta, maxiter=100):
    w_ext = w_ext.copy().astype(float)
    # Augment feature matrix with leading column of ones (x_0 = 1)
    X_ext = np.c_[np.ones(len(X)), X]
    
    for epoch in range(maxiter):
        for i in range(len(X_ext)):
            output = adaline_activation_ext(X_ext[i], w_ext)
            error = y[i] - output
            w_ext += eta * error * X_ext[i]
            
    return w_ext
```

## Testing

```python
w_ext_init = np.array([theta_init, w[0], w[1]])

w_ext = adaline_learning_weights(X, y, w_ext_init, eta, maxiter)

print(f"Final w:     [{w_ext[1]:.8f}, {w_ext[2]:.8f}]")
print(f"Final theta: {w_ext[0]:.8f}")
```
